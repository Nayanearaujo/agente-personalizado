## 1. Programação e lógica

Um computador executa instruções com enorme velocidade e precisão, mas não tem iniciativa: faz exatamente o que mandamos, na ordem em que mandamos. **Programar é escrever essas instruções de um jeito que a máquina consiga executar.**

### 1.1 O que é uma linguagem de programação

Imagine que você precisa dar instruções a alguém que só entende um idioma muito específico, com regras rígidas e nenhuma margem para interpretação. Uma **linguagem de programação** é esse idioma: um conjunto de palavras e regras que permite escrever instruções que o computador entende.

O processador (o “cérebro” do computador) só entende **código de máquina**, que são sequências de zeros e uns. Ninguém escreve zeros e uns à mão. Por isso foram criadas linguagens mais próximas do idioma humano, e um programa **tradutor** converte o que escrevemos para código de máquina. Existem dois tipos de tradutor, o compilador e o interpretador, que explicamos no Módulo 4.

### 1.2 Lógica e algoritmo

Antes de qualquer linguagem vem a lógica. Um **algoritmo** é uma sequência finita e ordenada de passos para resolver um problema. Pense numa receita de bolo: os ingredientes são a **entrada**, o modo de preparo é o **processamento** e o bolo pronto é a **saída**. Todo algoritmo tem essas três partes.

Exemplo: calcular a média das notas de um aluno. Em português, os passos são:

```text
1. Receber as notas do aluno          (entrada)
2. Somar as notas                      (processamento)
3. Dividir a soma pela quantidade      (processamento)
4. Mostrar a média                     (saída)
```

Agora o mesmo algoritmo em Python, em três células, um passo de cada vez.

**Passo 1: guardar as notas.** A linha abaixo cria uma **lista**, que é um conjunto de valores entre colchetes separados por vírgula. O nome `notas` passa a se referir a essa lista. Atenção: os números decimais usam **ponto**, e não vírgula.

````python
notas = [7.5, 8.0, 6.5]
````

**Passo 2: calcular a média.** `sum(notas)` soma todos os valores da lista (7,5 + 8,0 + 6,5 = 22,0). `len(notas)` conta quantos itens a lista tem (3). A barra `/` faz a divisão. O resultado fica guardado no nome `media`.

````python
media = sum(notas) / len(notas)
````

**Passo 3: mostrar o resultado.** `print(...)` exibe algo na tela. O `f` antes das aspas cria uma *f-string*: o que estiver entre chaves `{ }` é trocado pelo valor. O trecho `:.2f` pede duas casas decimais.

````python
print(f"Média: {media:.2f}")
````

> **Ideia central:** Toda lógica de programação se resume a três estruturas: **sequência** (um passo depois do outro), **decisão** (se isso, então aquilo) e **repetição** (faça várias vezes). Nesta Parte 1 trabalhamos a sequência e os dados. Decisão e repetição entram na Parte 2.

Por que lógica primeiro? Porque a sintaxe muda de linguagem para linguagem, mas a lógica permanece. Quem pensa bem o problema escreve o código em qualquer linguagem.

## 4. Características do Python

Para conhecer uma linguagem, os profissionais costumam classificá-la respondendo algumas perguntas. Neste módulo respondemos cada uma para o Python, uma de cada vez, partindo do zero:

1. É de **alto** ou de **baixo nível**?
2. É **compilada** ou **interpretada**?
3. Tem tipagem **dinâmica** ou **estática**? **Forte** ou **fraca**?
4. É **imperativa** ou **declarativa**?

### 4.1 Alto nível ou baixo nível?

Primeiro, um conceito: a **memória**. Quando o computador trabalha, guarda os dados temporariamente na memória (a RAM). Imagine uma enorme estante com milhões de gavetas numeradas. O número é o **endereço** da gaveta. Para guardar um dado, é preciso escolher gavetas livres, usá-las e, no fim, liberá-las.

**Nível** diz o quanto a linguagem está perto da máquina ou perto do jeito humano de pensar.

**Baixo nível** significa perto da máquina. Você mesmo cuida de detalhes como reservar espaço na memória, dizer de quanto espaço precisa e devolvê-lo ao final. Veja como seria, em C (uma linguagem de baixo nível), guardar três números inteiros:

```c
int *numeros = malloc(3 * sizeof(int));   // 1. reserva espaço na memória
numeros[0] = 10;                          // 2. guarda os valores
numeros[1] = 20;
numeros[2] = 30;
free(numeros);                            // 3. devolve o espaço
```

- **Linha 1:** `malloc` pede ao sistema um pedaço de memória, e **você** calcula o tamanho (3 números vezes o tamanho de cada inteiro).
- **Linhas 2 a 4:** você guarda os valores nas posições reservadas.
- **Linha 5:** `free` devolve o espaço. Se você esquecer, a memória fica ocupada sem necessidade (um *vazamento de memória*). Se usar uma posição errada, o programa pode travar ou corromper dados.

Em troca desse trabalho, você tem controle total e programas muito rápidos.

**Alto nível** significa perto do jeito humano de pensar. A linguagem esconde esses detalhes e cuida da memória por você. O mesmo exemplo em Python:

````python
numeros = [10, 20, 30]
print(numeros)
````

Uma linha. Você não reserva nem libera nada: o Python decide onde guardar e limpa sozinho quando o dado não é mais usado (esse faxineiro automático se chama *coletor de lixo*).

Uma analogia: dirigir um carro de **câmbio manual** (baixo nível) exige controlar a embreagem e trocar as marchas. Dá mais controle, mas dá mais trabalho e é fácil errar. O **câmbio automático** (alto nível) cuida disso por você, e você só pensa no destino.

| Nível | Exemplos | Você cuida da memória? |
|---|---|---|
| Muito baixo | Assembly (instruções do processador, uma a uma) | Sim, de tudo |
| Baixo | C | Sim, reserva e libera |
| Alto | Python, Java, JavaScript | Não, a linguagem cuida |

> **Na engenharia de dados:** O preço do alto nível é velocidade: um laço puro em Python é bem mais lento que o equivalente em C. Isso raramente pesa, porque bibliotecas como NumPy e pandas, e motores como o Spark, executam o trabalho pesado em código de baixo nível otimizado (C, C++, Rust, Java), enquanto você escreve em Python. **Python funciona como a cola**: descreve o que fazer e delega a execução para motores rápidos.

### 4.2 Compilada ou interpretada?

Lembra do tradutor do Módulo 1? Existem dois tipos:

- **Compilador:** traduz o programa **inteiro** de uma vez, antes de executar, e gera um arquivo executável. É como traduzir um livro inteiro antes de entregá-lo ao leitor.
- **Interpretador:** lê e executa o código **aos poucos**, linha a linha. É como um intérprete que traduz uma conversa em tempo real.

C é tipicamente compilada. **Python é interpretada**: você escreve e executa direto, sem etapa extra de compilação. Isso é ótimo para testar e aprender, e é por isso que o notebook funciona célula a célula.

> **Curiosidade:** Por dentro, o Python converte o seu código para um formato intermediário (o *bytecode*) e o executa numa máquina virtual. Para quem está começando basta saber: você escreve e roda.

### 4.3 Tipagem: dinâmica e forte

O **tipo** de um dado é a sua natureza: número inteiro, número decimal, texto e assim por diante (Módulo 6 detalha). A tipagem descreve como a linguagem lida com isso. São duas perguntas diferentes.

**Dinâmica ou estática?** Em linguagens estáticas, como C, você precisa declarar o tipo antes de usar (`int idade = 29;`). Em linguagens **dinâmicas**, como Python, não declara nada: o Python descobre o tipo pelo valor, e o mesmo nome pode apontar para valores de tipos diferentes ao longo do tempo. A função `type()` mostra o tipo de um valor.

````python
x = 10
print(type(x))
````

`int` significa número inteiro. Agora o mesmo nome `x` recebe um texto:

````python
x = "dez"
print(type(x))
````

`str` (de *string*) significa texto. O mesmo `x` mudou de tipo, e o Python aceitou sem reclamar. Isso é tipagem dinâmica.

**Forte ou fraca?** Em linguagem de tipagem **forte**, como Python, os tipos não se misturam em silêncio: somar um texto com um número gera um erro, em vez de a linguagem adivinhar o que você quis. Para mostrar o erro sem interromper o notebook, usamos `try` e `except`, que significam “tente executar isto; se der erro, faça aquilo”. Você verá isso com calma na Parte 2.

````python
try:
    resultado = "5" + 2
except TypeError as erro:
    print("Erro de tipo:", erro)
````

A mensagem diz que só dá para juntar texto com texto. Em JavaScript, uma linguagem de tipagem fraca, `"5" + 2` devolveria `"52"` sem avisar nada.

### 4.4 Imperativa ou declarativa?

Chamamos de **paradigma** o jeito de pensar e organizar o código. Dois paradigmas importantes para quem trabalha com dados são o imperativo e o declarativo. Uma analogia ajuda.

Imagine que você quer comer uma pizza. Uma opção é ir à cozinha e seguir uma receita: “abra a massa, espalhe o molho, ponha o queijo, leve ao forno por 15 minutos”. Você descreve **como** fazer, passo a passo. A outra opção é ligar para uma pizzaria e dizer: “quero uma pizza de calabresa”. Você descreve **o que** quer, e a cozinha decide como fazer.

- **Imperativo** é a receita: você ordena cada passo, na ordem. Descreve o *como*.
- **Declarativo** é o pedido: você declara o resultado que quer. Descreve o *quê*.

Vamos ver isso com um problema de dados: **somar apenas as vendas acima de 100**, numa lista de vendas com os valores 50, 120, 300, 80 e 150. A resposta esperada é 120 + 300 + 150 = 570.

**Estilo declarativo, em SQL.** Pedimos o resultado, sem dizer como chegar nele:

```sql
SELECT SUM(valor) FROM vendas WHERE valor > 100;
```

Leia em português: “some o valor, da tabela vendas, onde o valor for maior que 100”. Não dissemos como percorrer a tabela nem em que ordem. O banco de dados decide.

**Estilo imperativo, em Python.** Aqui escrevemos a receita. Em português, seria:

```text
1. Comece com total igual a zero
2. Para cada valor da lista
3.     se o valor for maior que 100
4.         some o valor ao total
5. No fim, mostre o total
```

Primeiro, guardamos a lista de valores:

````python
valores = [50, 120, 300, 80, 150]
````

Agora a receita em código. Você ainda não estudou `for` e `if` (são da Parte 2), então não precisa dominar: apenas compare com os passos acima. `for v in valores` significa “para cada `v` na lista `valores`”. `if v > 100` significa “se `v` for maior que 100”. O recuo (os espaços à esquerda) indica o que está dentro de cada passo.

````python
total = 0
for v in valores:
    if v > 100:
        total = total + v
print("Imperativo:", total)
````

Python também permite escrever o mesmo de forma declarativa, em uma linha. Lê-se: “some `v`, para cada `v` em `valores`, se `v` for maior que 100”. Descrevemos o que queremos, sem a receita passo a passo.

````python
print("Declarativo:", sum(v for v in valores if v > 100))
````

Mesmo resultado, dois estilos. Python é **multiparadigma**: aceita os dois (e outros, como orientação a objetos e programação funcional, que não veremos agora). Na prática, a maior parte do código Python é **imperativa**: você manda, passo a passo, o que o programa deve fazer.

> **Na engenharia de dados:** SQL e Spark são declarativos de verdade: você descreve a transformação e o motor decide como executá-la da forma mais eficiente. O pandas fica no meio do caminho: você descreve operações sobre colunas inteiras, sem escrever o laço. Já o Python imperativo faz a orquestração da pipeline: chama a API, lê o arquivo, trata o erro, dispara o próximo passo.

## 5. Variáveis

Uma variável é um **nome que referencia um valor** guardado na memória. Pense numa etiqueta colada no valor, e não numa caixa que guarda o valor dentro. Essa diferença explica comportamentos importantes mais adiante.

O sinal `=` é a **atribuição**. Ele não significa “igual” como na matemática: lê-se “o nome à esquerda passa a apontar para o valor à direita”. A célula abaixo cria quatro variáveis, cada uma com um tipo diferente de valor: um texto (entre aspas), um número inteiro, um decimal (com ponto) e um valor lógico (`True` ou `False`, sempre com a primeira letra maiúscula).

````python
nome_tabela = "clientes"
qtd_linhas = 1500
taxa_nulos = 0.032
ativa = True
````

Para ver o conteúdo das variáveis, usamos `print`. Quando damos vários valores separados por vírgula, o `print` os mostra separados por espaço.

````python
print(nome_tabela, qtd_linhas, taxa_nulos, ativa)
````

### 5.1 Regras e convenções de nomes

- Pode conter letras, números e sublinhado (`_`), mas **não pode começar com número**.
- **Diferencia maiúsculas de minúsculas:** `total` e `Total` são nomes diferentes.
- **Não pode ser palavra reservada** (`class`, `for`, `if`, `None`...).
- Não pode ter espaço nem hífen.
- Padrão PEP 8: **snake_case** (`qtd_linhas`) para variáveis e MAIÚSCULAS para constantes (`LIMITE_MAXIMO`).
- Prefira nomes descritivos: `qtd_linhas` comunica mais que `x`.
- Não use nomes de funções prontas (`list`, `sum`, `max`, `str`) como variável, pois você as perde no resto do código.

| Nome | Vale? | Motivo |
|---|---|---|
| `qtd_linhas` | Sim | Letras e sublinhado |
| `total2` | Sim | Número no fim é permitido |
| `2total` | Não | Começa com número |
| `nome-tabela` | Não | Hífen não é permitido (o Python lê como subtração) |
| `valor total` | Não | Espaço não é permitido |
| `class` | Não | Palavra reservada |

### 5.2 Reatribuição e atribuição múltipla

Uma variável pode receber um novo valor a qualquer momento. A etiqueta simplesmente passa a apontar para outro valor, e o antigo é descartado.

````python
contador = 1
print(contador)

contador = 2
print(contador)
````

Python permite atribuir vários nomes de uma vez. A linha `a, b = 10, 20` dá 10 para `a` e 20 para `b`. Isso permite **trocar valores** sem variável auxiliar: `a, b = b, a`.

````python
a, b = 10, 20
print(a, b)

a, b = b, a
print(a, b)
````

Também dá para colocar duas etiquetas no mesmo valor com `x = y = 0`.

````python
x = y = 0
print(x, y)
````

### 5.3 Variável é referência

Como a variável é uma etiqueta, duas etiquetas podem apontar para o mesmo objeto. Quando o objeto é **mutável** (pode ser alterado no lugar, como uma lista), mudar por uma etiqueta muda o que a outra enxerga. Essa é uma fonte clássica de bugs em tratamento de dados. Vamos ver em três passos.

**Passo 1.** Criamos uma lista e, depois, `lista_b = lista_a`. Isso **não copia** a lista: apenas cria uma segunda etiqueta para a mesma lista.

````python
lista_a = [1, 2, 3]
lista_b = lista_a
````

**Passo 2.** O método `.append(4)` adiciona o valor 4 no fim da lista. Mexemos pela etiqueta `lista_b`, mas olhe o que acontece com `lista_a`. O operador `is` pergunta se os dois nomes apontam para o mesmo objeto.

````python
lista_b.append(4)

print(lista_a)
print(lista_a is lista_b)
````

**Passo 3.** Para ter uma cópia de verdade, usamos `.copy()`. Agora `lista_c` é uma lista independente.

````python
lista_c = lista_a.copy()
lista_c.append(5)

print(lista_a)
print(lista_c)
````

## 6. Tipos de dados

Todo valor em Python tem um **tipo**, que define o que se pode fazer com ele. Por exemplo, dá para somar dois números, mas não faz sentido dividir um texto por outro. Os tipos básicos:

| Tipo | Exemplo | Uso em dados |
|---|---|---|
| `int` | `42` | Identificadores, contagens, quantidades |
| `float` | `3.14` | Medidas, taxas, valores com casas decimais |
| `str` | `"Ana"` | Nomes, códigos, datas em formato texto |
| `bool` | `True`, `False` | Indicadores (ativo, válido, duplicado) |
| `NoneType` | `None` | Ausência de valor (o nulo) |

### 6.1 Descobrindo o tipo de um valor

A função `type()` diz o tipo de qualquer valor. A função `isinstance(valor, tipo)` pergunta se o valor é daquele tipo e responde `True` ou `False`.

````python
print(type(42))
print(type(3.5))
print(type("a"))
print(type(True))
````

````python
print(isinstance(3.5, float))
print(isinstance(3.5, int))
````

### 6.2 Números: int e float

O `int` é o número inteiro, sem casas decimais, e **não tem limite de tamanho**. O operador `**` é a potência. Veja um número enorme:

````python
print(2 ** 100)
````

Para facilitar a leitura de números grandes, o Python aceita o sublinhado dentro do número:

````python
print(1_000_000)
````

O `float` é o número com casas decimais. Ele tem uma pegadinha. O computador guarda números em **binário** (zeros e uns), e alguns decimais, como 0,1, não têm representação exata em binário, do mesmo jeito que 1/3 não tem representação exata em decimal (0,333...). O resultado é uma pequena imprecisão:

````python
print(0.1 + 0.2)
````

Para dinheiro e outros casos em que a exatidão importa, existe o tipo `Decimal`, que precisa ser trazido com `import`. Note que o valor entra como texto, entre aspas:

````python
from decimal import Decimal
print(Decimal("0.1") + Decimal("0.2"))
````

A função `round(número, casas)` arredonda para a quantidade de casas indicada:

````python
print(round(3.14159, 2))
````

> **Atenção:** Em dados financeiros, nunca compare valores `float` com `==` nem some centavos em `float` sem cuidado. Use `Decimal` ou trabalhe em centavos como `int`.

### 6.3 Strings (texto)

Uma string é uma sequência de caracteres entre aspas simples ou duplas. Para textos de várias linhas usam-se aspas triplas. Strings são **imutáveis**: nenhum método altera o texto original, todos devolvem um texto novo.

````python
texto = "  Engenharia de Dados  "
print(texto)
````

Um **método** é uma função que pertence a um tipo e é chamada com um ponto depois do valor. `.upper()` devolve o texto em maiúsculas, `.lower()` em minúsculas e `.strip()` remove os espaços das pontas. Note que `texto` continua igual: guardamos o resultado de `.strip()` numa variável nova.

````python
print(texto.upper())
print(texto.lower())
limpo = texto.strip()
print(limpo)
````

`len()` conta quantos caracteres o texto tem (espaços contam).

````python
print(len(limpo))
````

**Indexação.** Cada caractere ocupa uma posição, chamada de **índice**. A contagem começa em **0**, e índices negativos contam a partir do fim. Veja a palavra `Dados`:

```text
 D   a   d   o   s
 0   1   2   3   4      (índices positivos: começam em 0)
-5  -4  -3  -2  -1      (índices negativos: contam do fim)
```

````python
palavra = "Dados"
print(palavra[0])
print(palavra[-1])
````

**Fatiamento.** `[início:fim]` extrai um pedaço do texto. O índice do início **entra**, o do fim **não entra**. Por isso `[1:4]` pega as posições 1, 2 e 3. Se omitir um dos lados, vai até a ponta.

````python
print(palavra[1:4])
print(palavra[:2])
print(palavra[2:])
````

Outros métodos muito usados: `.replace(antigo, novo)` troca um trecho por outro, `.startswith(trecho)` pergunta se o texto começa com aquele trecho, e o operador `in` pergunta se um trecho está dentro do texto.

````python
print(limpo.replace("Dados", "IA"))
print(limpo.startswith("Eng"))
print("Dados" in limpo)
````

`.split(separador)` quebra um texto em pedaços e devolve uma lista. `.join()` faz o caminho inverso. Os dois são a base para tratar linhas de arquivos como CSV, em que os campos vêm separados por um símbolo (aqui, o ponto e vírgula).

````python
linha = "101;Ana;São Paulo;1500.50"
campos = linha.split(";")
print(campos)
````

````python
print(" | ".join(campos))
````

As **f-strings** (texto com `f` antes das aspas) inserem valores dentro do texto e controlam a formatação. Dentro das chaves, `:.1%` mostra como porcentagem com uma casa, e `:.2f` mostra duas casas decimais.

````python
tabela = "vendas"
linhas = 12500
nulos = 0.0456
print(f"A tabela {tabela} tem {linhas} linhas e {nulos:.1%} de nulos.")
print(f"Valor: {1234.5:.2f}")
````

### 6.4 Booleanos e valores “verdadeiros”

O tipo `bool` só tem dois valores: `True` e `False`. Além disso, em Python qualquer valor pode ser avaliado como verdadeiro ou falso com `bool()`. São considerados **falsos**: o número zero, o texto vazio, a coleção vazia e `None`. Todo o resto é verdadeiro.

````python
print(bool(0), bool(""), bool([]), bool(None))
print(bool(1), bool("a"), bool([0]))
````

Cuidado com uma pegadinha: o texto `"False"` não está vazio, então é verdadeiro.

````python
print(bool("False"))
````

### 6.5 None: o nulo

`None` representa a **ausência de valor**. Não é zero e não é texto vazio. Em engenharia de dados é o equivalente ao `NULL` do SQL. Para testar, compara-se com `is None`.

````python
resultado = None
print(resultado is None)
print(type(resultado))
print(resultado == 0, resultado == "")
````

### 6.6 Convertendo entre tipos

As funções `int()`, `float()` e `str()` convertem um valor para outro tipo. Note que `int()` aplicado a um decimal **trunca** (corta as casas), não arredonda.

````python
print(int("42") + 1)
print(float("3.5"))
print(str(42) + " itens")
print(int(3.99))
````

Quando a conversão é impossível, o Python gera um `ValueError`. Aqui usamos `try` e `except` de novo para mostrar a mensagem sem interromper o notebook.

````python
try:
    int("abc")
except ValueError as erro:
    print("Erro:", erro)
````

> **Atenção:** Dados que chegam de arquivos, formulários e APIs costumam vir como **texto**, mesmo quando parecem número. A função `input()`, que lê o que o usuário digita, também sempre devolve `str`. Converter para o tipo certo é uma das primeiras tarefas de qualquer pipeline.

### 6.7 Mutável ou imutável

Um valor **imutável** não pode ser alterado depois de criado (qualquer mudança cria um valor novo). Um **mutável** pode ser alterado no lugar. Essa classificação importa para o Módulo 8.

| Imutáveis | Mutáveis |
|---|---|
| `int`, `float`, `str`, `bool`, `tuple`, `None` | `list`, `dict`, `set` |

## 7. Operadores

Operadores são os símbolos que realizam operações sobre valores.

| Grupo | Operadores | Observação |
|---|---|---|
| Aritméticos | `+  -  *  /  //  %  **` | `/` sempre devolve `float`. `//` é divisão inteira. `%` é o resto |
| Comparação | `==  !=  <  >  <=  >=` | Resultado é `True` ou `False` |
| Lógicos | `and  or  not` | Combinam condições |
| Pertinência | `in  not in` | Verifica se um item está numa sequência |
| Identidade | `is  is not` | Verifica se é o mesmo objeto (use com `None`) |
| Atribuição composta | `+=  -=  *=  /=` | `x += 1` equivale a `x = x + 1` |

### 7.1 Aritméticos

Cada linha abaixo mostra o cálculo e o resultado. Observe as três divisões: `7 / 2` dá 3.5 (decimal), `7 // 2` dá 3 (só a parte inteira) e `7 % 2` dá 1 (o resto da divisão).

````python
print("7 + 2  =", 7 + 2)
print("7 - 2  =", 7 - 2)
print("7 * 2  =", 7 * 2)
print("7 / 2  =", 7 / 2)
print("7 // 2 =", 7 // 2)
print("7 % 2  =", 7 % 2)
print("7 ** 2 =", 7 ** 2)
````

### 7.2 Comparação

Comparam dois valores e respondem `True` ou `False`. Atenção: `=` atribui, e `==` compara.

````python
print(5 == 5.0)
print(5 != 6)
print(3 > 2)
print("a" < "b")
````

### 7.3 Lógicos

`and` é verdadeiro só se **os dois** lados forem. `or` é verdadeiro se **pelo menos um** for. `not` inverte.

````python
print(True and False)
print(True or False)
print(not True)
````

### 7.4 Pertinência

`in` pergunta se um item está dentro de uma lista ou de um texto. `not in` pergunta o contrário.

````python
print(3 in [1, 2, 3])
print("x" not in "abc")
````

### 7.5 Atribuição composta

São atalhos para atualizar uma variável usando o próprio valor dela. `total += 50` é o mesmo que `total = total + 50`.

````python
total = 100
total += 50
print(total)

total *= 2
print(total)
````

> **Precedência:** Python respeita a ordem: `**`, depois `* / // %`, depois `+ -`, depois comparações, depois `not`, `and`, `or`. Na dúvida, use parênteses. Eles não custam nada e evitam erros.

## 8. Coleções: list, tuple, dict e set

Dados raramente aparecem sozinhos. Uma **coleção** guarda vários valores sob um mesmo nome. Python tem quatro coleções principais:

| Tipo | Sintaxe | Ordenado | Mutável | Duplicatas | Uso típico |
|---|---|---|---|---|---|
| `list` | `[1, 2, 3]` | Sim | Sim | Aceita | Sequência de registros |
| `tuple` | `(1, 2, 3)` | Sim | Não | Aceita | Registro fixo, chave composta |
| `dict` | `{"id": 1}` | Sim (inserção) | Sim | Chaves únicas | Registro, linha, JSON |
| `set` | `{1, 2, 3}` | Não | Sim | Não aceita | Valores distintos |

### 8.1 list

Uma **lista** é uma sequência ordenada e mutável, entre colchetes. Os itens são acessados pelo índice, começando em 0, como nas strings.

````python
ids = [101, 102, 103]
print(ids)
````

Para alterar a lista: `.append(x)` adiciona no fim, `.insert(posição, x)` adiciona numa posição e `.remove(x)` remove pelo valor.

````python
ids.append(104)
ids.insert(0, 100)
ids.remove(102)
print(ids)
````

Para consultar: `len()` conta os itens, `ids[0]` é o primeiro, `ids[-1]` é o último e `ids[1:3]` é uma fatia (posições 1 e 2).

````python
print(len(ids))
print(ids[0], ids[-1])
print(ids[1:3])
````

`sorted()` devolve uma lista ordenada (sem alterar a original; `reverse=True` inverte). `sum()`, `min()` e `max()` calculam a soma, o menor e o maior.

````python
print(sorted(ids, reverse=True))
print(sum(ids), min(ids), max(ids))
````

### 8.2 tuple

A **tupla** é parecida com a lista, mas **imutável**: depois de criada, não muda. Usa parênteses. É boa para dados que não devem mudar, como coordenadas. Dá para **desempacotar** seus valores em variáveis separadas numa única linha.

````python
coordenada = (-19.92, -43.94)
lat, lon = coordenada
print(lat, lon)
````

### 8.3 dict

O **dicionário** guarda pares **chave: valor**, entre chaves. Em vez de acessar por posição, você acessa pelo nome da chave. É a estrutura mais importante para quem trabalha com dados em Python, porque é a cara de um registro e de um JSON. Desde o Python 3.7 a ordem de inserção é mantida.

````python
cliente = {"id": 1, "nome": "Ana", "cidade": "Belo Horizonte"}
print(cliente["nome"])
````

Para adicionar uma chave nova ou alterar o valor de uma existente, usa-se a mesma sintaxe de atribuição:

````python
cliente["email"] = "ana@exemplo.com"
cliente["cidade"] = "Contagem"
print(cliente)
````

Acessar uma chave que não existe com `cliente["telefone"]` gera `KeyError`. O método `.get(chave, padrão)` evita o erro e devolve o valor padrão. Isso é comum em dados vindos de APIs, em que campos podem faltar.

````python
print(cliente.get("telefone", "não informado"))
````

`.keys()` lista as chaves e `.values()` lista os valores. Envolvemos em `list()` para exibir como lista.

````python
print(list(cliente.keys()))
print(list(cliente.values()))
````

### 8.4 set

O **conjunto** guarda valores **únicos**, sem ordem. Ao transformar uma lista em `set`, as duplicatas somem automaticamente.

````python
cidades = ["BH", "SP", "BH", "RJ", "SP"]
unicas = set(cidades)
print(len(unicas))
print(sorted(unicas))
````

Conjuntos permitem as operações de conjuntos da matemática: `|` é a união (todos os itens), `&` é a interseção (só os que estão nos dois) e `-` é a diferença (os que estão no primeiro e não no segundo). Usamos `sorted()` só para exibir em ordem.

````python
a = {1, 2, 3}
b = {3, 4}
print(sorted(a | b))
print(sorted(a & b))
print(sorted(a - b))
````

> **Na engenharia de dados:** `dict` é um registro ou um JSON. Uma `list` de `dict` é uma tabela ou a resposta de uma API. `set` é o `DISTINCT` do SQL. `tuple` serve de chave composta. Entender essa correspondência é o que torna pandas e PySpark naturais depois.

## 9. Mini projeto: relatório de vendas

Vamos juntar tudo: uma lista de dicionários (nossa tabela), conversões de tipo, coleções, operadores e f-strings, para produzir um relatório. Cada passo é uma célula.

### Passo 1: os dados

Cada venda é um dicionário, e a lista `vendas` reúne todas. Para aprender: `len(vendas)` conta os registros, `vendas[0]` é o primeiro e `vendas[-1]["cliente"]` pega o cliente do último.

````python
vendas = [
    {"id": 1, "cliente": "Ana",   "produto": "Notebook", "valor": 4500.00},
    {"id": 2, "cliente": "Bruno", "produto": "Mouse",    "valor": 89.90},
    {"id": 3, "cliente": "Ana",   "produto": "Teclado",  "valor": 250.00},
    {"id": 4, "cliente": "Carla", "produto": "Monitor",  "valor": 1200.00},
    {"id": 5, "cliente": "Bruno", "produto": "Notebook", "valor": 5200.00},
    {"id": 6, "cliente": "Diego", "produto": "Mouse",    "valor": 79.90},
]
````

````python
print(len(vendas))
print(vendas[0])
print(vendas[-1]["cliente"])
````

### Passo 2: extraindo colunas

Para somar os valores, precisamos de uma lista só com eles. A linha `[v["valor"] for v in vendas]` lê-se “para cada venda `v` na lista, pegue o valor”. Usando chaves em vez de colchetes, `{v["cliente"] for v in vendas}`, o resultado é um `set`, ou seja, os clientes sem repetição.

````python
valores = [v["valor"] for v in vendas]
clientes = {v["cliente"] for v in vendas}
````

````python
print(valores)
print(sorted(clientes))
````

> **Prévia:** Essa construção é chamada de *comprehension* e é o jeito Python de extrair uma coluna. Será explicada com calma na Parte 2.

### Passo 3: métricas

Agora usamos as funções que já conhecemos: `sum`, `len`, `max` e `min`. O ticket médio é o faturamento dividido pela quantidade de vendas.

````python
total = sum(valores)
ticket_medio = total / len(valores)
maior = max(valores)
menor = min(valores)

print(round(total, 2), round(ticket_medio, 2), maior, menor)
````

### Passo 4: o relatório

O formato numérico padrão do Python usa vírgula para milhar e ponto para decimal (11,319.80). Para o padrão brasileiro (11.319,80), trocamos os símbolos. Como vamos repetir isso várias vezes, empacotamos o código numa **função**, usando `def`. Uma função é um bloco de código com nome, que podemos chamar quando quisermos. Aqui, `brl` recebe um valor e devolve o texto formatado. Esta célula apenas define a função, e por isso não mostra nada.

````python
def brl(valor):
    texto = f"{valor:,.2f}"
    texto = texto.replace(",", "X").replace(".", ",").replace("X", ".")
    return f"R$ {texto}"
````

> **Prévia:** Funções são tema da Parte 2. Por ora, basta entender que `brl(5200)` devolve `R$ 5.200,00`.

Agora o relatório. `"=" * 40` repete o caractere 40 vezes, para desenhar uma linha. `", ".join(...)` junta os nomes dos clientes separados por vírgula.

````python
print("=" * 40)
print("RELATÓRIO DE VENDAS")
print("=" * 40)
print(f"Registros:          {len(vendas)}")
print(f"Faturamento total:  {brl(total)}")
print(f"Ticket médio:       {brl(ticket_medio)}")
print(f"Maior venda:        {brl(maior)}")
print(f"Menor venda:        {brl(menor)}")
print(f"Clientes distintos: {len(clientes)} ({', '.join(sorted(clientes))})")
````

### Passo 5: dict é JSON

O módulo `json` converte dicionários em texto JSON. O argumento `indent=2` deixa o texto bem recuado, e `ensure_ascii=False` mantém os acentos.

````python
import json

print(json.dumps(vendas[0], indent=2, ensure_ascii=False))
````

A semelhança não é coincidência. O formato JSON, usado por praticamente toda API, foi desenhado sobre a mesma ideia de chave e valor. Na Parte 2 vamos ler esses dados de arquivos CSV e JSON em vez de digitá-los.
